```python
import pandas as pd
 
FILE = "Supply chain logisitcs problem.xlsx"
sheets = pd.read_excel(FILE, sheet_name=None)   # dict: {sheet_name: DataFrame}
print("Sheets found:", list(sheets.keys()))
```

```python
for name, df in sheets.items():
    print(f"\n=== {name} ===")
    print("Shape:", df.shape)
    print("Columns:", list(df.columns))
    null_counts = df.isna().sum()
    nulls = null_counts[null_counts > 0]
    if len(nulls):
        print("Columns with nulls:")
        print(nulls)
    else:
        print("No nulls.")
```

```python
for name, df in sheets.items():
    messy_cols = [c for c in df.columns if c != c.strip()]
    if messy_cols:
        print(f"{name}: column name(s) with extra whitespace -> {messy_cols}")
```

```python
for name, df in sheets.items():
    out_name = f"raw_{name.lower()}.csv"
    df.to_csv(out_name, index=False)
    print("Saved", out_name)
```
