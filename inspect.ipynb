```python
import os
import pandas as pd
 
files = [f for f in os.listdir(".") if not f.startswith(".") and not f.endswith(".py")]
print("Files found in this folder:")
for f in files:
    print(" -", f)
```

```python
for f in files:
    if f.lower().endswith((".xlsx", ".xls")):
        print(f"\n=== {f} ===")
        xls = pd.ExcelFile(f)
        for sheet in xls.sheet_names:
            df = xls.parse(sheet, nrows=5)   # peek at the first 5 rows only
            print(f"  Sheet '{sheet}': columns = {list(df.columns)}")
```

```python
for f in files:
    if f.lower().endswith(".csv"):
        df = pd.read_csv(f)
        print(f"\n=== {f} ===")
        print(f"  Shape: {df.shape}")
        print(f"  Columns: {list(df.columns)}")
```
